```python
import pandas as pd

data=pd.read_csv("customer_shopping_behavior (1).csv")
data.head()
```

```python
data.info()
```

```python
data.describe(include='all')
```

```python
round(data.isnull().sum()/len(data)*100,1)
```

```python
data.groupby("Category")["Review Rating"].mean()
```

```python
data["Review Rating"]=data["Review Rating"].fillna(data.groupby("Category")["Review Rating"].transform("median"))
```

```python
round(data.isnull().sum()/len(data)*100,1)
```

```python
data.columns=data.columns.str.lower()
```

```python
data.columns
```

```python
data.columns=data.columns.str.replace(" ","_")
```

```python
data.columns
```

```python
data=data.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
data.columns
```

```python
labels=["Young Adult","Adult","Middle-aged","Senior"]
data["age_groups"]=pd.qcut(data["age"],4,labels)
```

```python
data[["age","age_groups"]].head(10)
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

data["purchase_frequency_days"]=data["frequency_of_purchases"].map(frequency_mapping)
```

```python
data[["previous_purchases","frequency_of_purchases"]].head(10)
```

```python
(data['discount_applied']==data['promo_code_used']).all()
```

```python
data=data.drop("promo_code_used",axis=1)
```
